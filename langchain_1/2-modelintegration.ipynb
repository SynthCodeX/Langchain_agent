## Models integration 

```python
import os 
from dotenv import load_dotenv
load_dotenv()

os.environ["ANTHROPIC_API_KEY"] = os.getenv("ANTHROPIC_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model_name = "claude-haiku-4-5-20251001"


model = init_chat_model("claude-haiku-4-5-20251001")
model
```

```python
response = model.invoke("Hello how are you?")
response.content
```

### Streaming and Batch

```python
## Streaming
model.invoke("What is the latest trend in AI?")
```

```python
for chunk in model.stream("what is the latest trend in AI?"):
    print(chunk.content, end=" | ", flush=True)
```

```python
## Batch 

response = model.batch([
    "What is the latest trend in AI?",
    "How it is affecting Human Life?",
    "what are the carrer opportunities in AI?"
])

for response in response:
    print(response.content)
```
